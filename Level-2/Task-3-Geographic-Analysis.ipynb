```python
import  pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("../dataset/Dataset.csv")
```

```python
data.columns
```

```python
data =data[[ 'Longitude', 'Latitude']]
```

```python
data =data.dropna()
```

```python
data = data[(data['Latitude'] !=0) & (data["Longitude"] !=0)]
```

```python
plt.figure(figsize = (8,6))
plt.scatter(data["Longitude"],data["Latitude"],alpha = 0.5)
plt.xlabel("Lomgitude")
plt.ylabel("Latitude")
plt.title("GEOGRAPHIC DISTRIBUTION OF RESTAURANTS")
plt.show()
```

### Analysis – Geographic Distribution of Restaurants

Restaurant locations were visualized using latitude and longitude coordinates.
The scatter plot shows that restaurants are not evenly distributed geographically.
There are clear clusters in certain regions, indicating higher restaurant density
in urban or high-activity areas, while other regions show sparse distribution.
