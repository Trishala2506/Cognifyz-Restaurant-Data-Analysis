```python
import pandas as pd
```

```python
data = pd.read_csv("../dataset/Dataset.csv")
```

```python
data.columns
```

```python
data = data[["Aggregate rating","Votes"]]
```

```python
data = data.dropna()
```

```python
data.isnull().sum()
```

```python
bins = [0,1,2,3,4,5]
labels = ['0-1','1-2','2-3','3-4','4-5']
data[" Rating Range"] = pd.cut(data["Aggregate rating"],bins=bins,labels=labels,include_lowest=True)
```

```python
rating_distribution = data[" Rating Range"].value_counts()
```

```python
rating_distribution
```

```python
most_common_rating =   rating_distribution.idxmax()
```

```python
most_common_rating
```

```python
avg_votes = data["Votes"].mean()
```
