```python
import pandas as pd
```

```python
data =pd.read_csv("../dataset/Dataset.csv")
```

```python
data.columns
```

```python
chain_counts = data['Restaurant Name'].value_counts()
restaurant_chains = chain_counts[chain_counts > 1]
restaurant_chains.head()
```

```python
chain_names = restaurant_chains.index

chain_data = data[data['Restaurant Name'].isin(chain_names)]
avg_chain_rating = chain_data.groupby('Restaurant Name')['Aggregate rating'].mean()
```

```python
outlet_count = chain_data['Restaurant Name'].value_counts()
```

```python
avg_chain_votes = chain_data.groupby('Restaurant Name')['Votes'].mean()
```

```python
chain_analysis = pd.DataFrame({
    'Outlets': outlet_count,
    'Average Rating': avg_chain_rating,
    'Average Votes': avg_chain_votes
})

chain_analysis = chain_analysis.sort_values(by='Outlets', ascending=False)
chain_analysis.head(10)
```

The dataset contains several restaurant chains, identified by multiple occurrences of the same restaurant name.
Popular chains tend to have a higher number of outlets and receive more customer votes.
Some chains also maintain consistently high average ratings, indicating strong brand quality and customer trust.
