```python
rate_data = data.dropna(subset = ["Aggregate rating"])
```

```python
avg_rate_Online = rate_data.groupby('Has Online delivery')[ 'Aggregate rating'].mean()
```

```python
avg_rate_Online
```

```python
final_Table = pd.DataFrame({
    "Online Delivery": avg_rate_Online.index,
    "Average Rating": avg_rate_Online.values
})
final_Table
```
