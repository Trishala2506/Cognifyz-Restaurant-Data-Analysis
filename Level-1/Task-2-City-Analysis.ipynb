```python
import pandas as pd
```

```python
data = pd.read_csv("../dataset/Dataset.csv")
```

```python
data.columns
```

```python
city_count = data["City"].value_counts()
```

```python
city_count.head()
```

```python
top_city_restaurants = city_count.idxmax()
```

```python
top_city_restaurants
```

```python
clear_rating =  data.dropna(subset=["Aggregate rating"])
```

```python
CITY_AVG_Rating = clear_rating.groupby("City")["Aggregate rating"].mean()
```

```python
CITY_AVG_Rating.head()
```

```python
top_city_BY_Ratings = CITY_AVG_Rating.idxmax()
```

```python
top_city_BY_Ratings
```

```python
final_City_Result = pd.DataFrame({
    "City":CITY_AVG_Rating.index,"AVG_Rating":CITY_AVG_Rating.values})
```
