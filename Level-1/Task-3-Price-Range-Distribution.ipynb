```python
import pandas as pd
```

```python
data = pd.read_csv("../dataset/Dataset.csv")
```

```python
data.columns
```

```python
price_count = data["Price range"].value_counts().sort_index()
```

```python
price_count 
```

```python
import matplotlib.pyplot as plt
```

```python
price_count.plot(kind="bar")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.title("Price Distribution of Restaurants")
```

```python
total_Resto = price_count.sum()
```

```python
price_perc = (price_count /total_Resto)*100
```

```python
price_perc
```

```python
price_range_table = pd.DataFrame({
    "Price_Range": price_count.index,
    "Restaurant_Count": price_count.values,
    "Percentage": price_perc.values
})
price_range_table
```
