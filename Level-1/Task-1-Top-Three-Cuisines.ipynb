```python
import pandas as pd
```

```python
data = pd.read_csv("../dataset/Dataset.csv")
```

```python
data.columns
```

```python
clean_data = data.dropna(subset=["Cuisines"])
```

```python
cuisine_count= clean_data["Cuisines"].value_counts()
```

```python
t3_cuisines = cuisine_count.head(3)
```

```python
t3_cuisines
```

```python
total_restaurant = len(clean_data)
```

```python
total_restaurant
```

```python
percentage = (t3_cuisines / total_restaurant)* 100
```

```python
percentage
```

```python
end_result = pd.DataFrame({"Cuisine":t3_cuisines.index,"Restaurant_Count":t3_cuisines.values,"percentage":percentage.values})
```
